# Basic usage

This notebook is a starting point for using GRAVITYp in a Jupyter environment.

```python
import gravityp

print(f"gravityp version: {gravityp.__version__}")
```

Schwarzachild metric:

$$ ds^2 = -\left(1-\frac{2M}{r}\right)dt^2 + \left(1-\frac{2M}{r}\right)^{-1}dr^2 + r^2d\theta^2 + r^2\sin^2\theta d\varphi^2 $$

```python
def g_rr(r: float) -> float:
    """Radial function g^{rr} of the Schwarzschild metric in units of M=1"""
    return 1 - 2./r

def g_thth(r: float) -> float:
    """Areal radius squared g^{theta theta} of the Schwarzschild in units of M=1"""
    return r**2
```

```python
from gravityp import CompactObject

# CompactObject instance codifying Schwarzschild metric and required params
schw = CompactObject(
    r_phs=[3.,],        # Photon sphere at r=3M
    inner_edge=2.,      # Inner edge of disk at horizon r=2M
    radial=g_rr,        # Above defined radial function
    areal=g_thth,       # Above defined areal radius squared
)
```

```python
import numpy as np
from gravityp import make_geodesics_plot

impact_parameters = np.arange(0,10,0.2)
make_geodesics_plot(impact_parameters, schw, figsize=(4,4))
```
